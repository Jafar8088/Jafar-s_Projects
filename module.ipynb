# Libraries

##### library is collection of modules and packages that provoide a prewritten functionality for your program. Libraries are typically larger and more features rich than packages or module.

To leverage powerfull tools developed by the community

# Python PIP 

PIP stands for "Pip Install Packages", and it is the package manager for python that allows you install, update, and manage python libraries (packages) from the Python Package Index (PyPI)

When you use the "pip install package_name" it connects to Python Package Index(PyPI) online. then download the specified libraries or packages then install it into your python environment.

Ex : install pandas to work on dataframe.
