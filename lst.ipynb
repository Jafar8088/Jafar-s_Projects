# LIST METHODS

python provids several builtin methos to operate and modify on a list.

| **Method**    | **Syntax**                                           | **What it does (Simple Words)**                                                        | **Example**                         | **Output**                           |
| ------------- | ---------------------------------------------------- | -------------------------------------------------------------------------------------- | ----------------------------------- | ------------------------------------ |
| **append()**  | `list_name.append(item)`                             | Adds **one item** at the end of the list.                                              | `l=[10,20]`<br>`l.append(30)`       | `[10,20,30]`                         |
| **extend()**  | `list_name.extend(iterable)`                         | Adds **multiple items** from another list (or iterable).                               | `l=[10,20]`<br>`l.extend([30,40])`  | `[10,20,30,40]`                      |
| **insert()**  | `list_name.insert(index, item)`                      | Adds an item at a **specific position**.                                               | `l=[10,20]`<br>`l.insert(1,15)`     | `[10,15,20]`                         |
| **remove()**  | `list_name.remove(value)`                            | Removes the **first matching value**.                                                  | `l=[10,20,20,30]`<br>`l.remove(20)` | `[10,20,30]`                         |
| **pop()**     | `list_name.pop(index)`                               | Removes and returns an item by **index**. If no index is given, removes the last item. | `l=[10,20,30]`<br>`l.pop()`         | Returns `30`, list becomes `[10,20]` |
| **clear()**   | `list_name.clear()`                                  | Removes **all items** from the list.                                                   | `l=[10,20]`<br>`l.clear()`          | `[]`                                 |
| **index()**   | `list_name.index(value)`                             | Finds the **position (index)** of an item.                                             | `l=[10,20,30]`<br>`l.index(20)`     | `1`                                  |
| **count()**   | `list_name.count(value)`                             | Counts how many times an item appears.                                                 | `l=[10,20,20,30]`<br>`l.count(20)`  | `2`                                  |
| **sort()**    | `list_name.sort()`<br>`list_name.sort(reverse=True)` | Arranges the list in ascending order by default.                                       | `l=[30,10,20]`<br>`l.sort()`        | `[10,20,30]`                         |
| **reverse()** | `list_name.reverse()`                                | Reverses the order of the list.                                                        | `l=[10,20,30]`<br>`l.reverse()`     | `[30,20,10]`                         |
| **copy()**    | `new_list = list_name.copy()`                        | Creates a copy of the list.                                                            | `l=[10,20]`<br>`new=l.copy()`       | `new=[10,20]`                        |

# flatten a nested list using list comprehension method

```python
nested_list = [[1,2],[3,4],[5,6]]
print('This is Nested List Format')
print(nested_list)
```

# Method 1

```python
nested_list = [[1,2],[3,4],[5,6]]
result = [item for sublist in nested_list for item in sublist]
print(nested_list)
print(result)

# first_itteration -> [1,2] -> 1,2
# second_itteration -> [3,4] -> 3,4
# third_itteration -> [5,6] -> 5,6
```

"This is used to flatten a nested list. Python first picks one sublist like [1,2], then takes each item 1 and 2. Next, it picks [3,4] and takes 3 and 4. Finally, it picks [5,6] and takes 5 and 6. All these items are added to a new list, giving the final output [1,2,3,4,5,6]."

The above output work like 

Python picks one sublist at a time ([1,2], [3,4], [5,6]). Then, from each sublist, it picks each item (1, 2, 3, 4, 5, 6) and adds it to the new list. The final output is a single flat list: [1,2,3,4,5,6]."

# Method 2

```python
nested_list = [[1,2],[3,4],[5,6]]
def flatten_list(lst): # make or define one function and give a parameter
    return [item for sublist in lst for item in sublist] #use comprehension list
final_list = flatten_list(nested_list)
print(nested_list)
print(final_list)
```
