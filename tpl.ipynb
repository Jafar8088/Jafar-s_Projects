# TUPPLES

##### Tupple is a collection of data items in python that is orderes , unchangable(immutable), and allow duplicate. And tupples are used to store multiple data items in a single variable.

```python
my_tuple = (1,2,3)
print(my_tuple)
print(type(my_tuple))
```

# Create tuples

### Method 1 : Using Paranthesis

```python
my_tuple = (1,2,3)
print(my_tuple)
print(type(my_tuple))
```

### creating tuples using mixed values

```python
mixed_tuple = (1,"Jafar",True,3.14)
print(mixed_tuple)
print(type(mixed_tuple))
```

### Method 2 : without using Paranthesis

```python
tuple1 = 1,2,3
print(tuple1)
print(type(tuple1))
```

### Method 3 : Using tuple constructor

```python
tuple2 = tuple((1,5,9)) #tuple constructors
print(tuple2)
print(type(tuple2))
```

# Convert the list into tuple

```python
list1 = [1,2,3]
new_tuple = tuple(list1)
print(list1)
print(type(list1))
print(new_tuple)
print(type(new_tuple))
```

# create a single element

```python
a = ("a") # currently this is a string
print(a)
print(type(a))
```

```python
a = ("a",) # put , comma to make tuple
print(a)
print(type(a))
```

# TUPPLE METHODS

| Method    | Syntax               | Example                    | Output |
| --------- | -------------------- | -------------------------- | ------ |
| `count()` | `tuple.count(value)` | `("a","b","a").count("a")` | `2`    |
| `index()` | `tuple.index(value)` | `("a","b","c").index("b")` | `1`    |


Easy way to remember:

count() → "How many times?"
index() → "At which position?"

# TUPLE FUNCTIONS

| Function   | Syntax      | Example         | Output      | Description                                                                                          |
| ---------- | ----------- | --------------- | ----------- | ---------------------------------------------------------------------------------------------------- |
| `len()`    | `len(t)`    | `t=(1,2,3,4,5)` | `5`         | Returns the **total number of elements** present in the tuple.                                       |
| `sorted()` | `sorted(t)` | `t=(5,2,8,1)`   | `[1,2,5,8]` | Returns a **new list with elements arranged in ascending order** (original tuple remains unchanged). |
| `sum()`    | `sum(t)`    | `t=(10,20,30)`  | `60`        | Calculates and returns the **sum of all numeric values** in the tuple.                               |
| `min()`    | `min(t)`    | `t=(7,2,9,4)`   | `2`         | Finds and returns the **smallest element** in the tuple.                                             |
| `max()`    | `max(t)`    | `t=(7,2,9,4)`   | `9`         | Finds and returns the **largest element** in the tuple.                                              |

# PACKING AND UNPACKING TUPLES

# PACKED TUPPLES

a . Packing : Packing is the process of putting multiple values is a single tuple.


```python
a = 'Jafar'
b = 21
c = 'Engineer'
print(a,b,c)
print(type(a))
print(type(b))
print(type(c))
```

```python
a = 'Sadiq'
b = 22
c = 'Analyst'
tuple_pack = a,b,c
print(tuple_pack)
print(type(tuple_pack))
```

# UNPACKED TUPLE

```python
name,age,profession = tuple_pack
print(name)
print(age)
print(profession)
```

# MODIFIYING TUPPLE - IMMUTABLE

Once a tupple is created, you cannot modify its elements that means you cannot add, remove or change its item.

```python
numbers = (10,20,30)
print(numbers)
print(type(numbers))
```

```python
numbers[0] = 99 # Error
```

# How to mutate / modify tuple

##### Step 1 : Convert the tupple to list using list constructor

```python
numbers = (10,20,30)
print(numbers)
print(type(numbers))
```

```python
tuple_numbers = (10,20,30)
list_numbers = list(tuple_numbers)
print(list_numbers)
print(type(list_numbers))
```

##### Step 2 : Replace the number to the specified position that you want or you can use append(), remove(), etc...,

```python
list_numbers[1] = 100
print(list_numbers)
```

##### Step 3 : Convert the list to tupple using tupple constructor

```python
tuple_numbers = tuple(list_numbers)
print(tuple_numbers)
print(type(tuple_numbers))
```

##### Using Tuples as Keys in Dictionaries

Since tuples are immutable and hashable, they can be used as keys in dictionaries, unlike lists.

```python
location_data = {
    (40.7128, -74.0060): "New York City",
    (34.0522, -118.2437): "Los Angeles"
}

print(location_data[(40.7128, -74.0060)])   # Output: New York City
```
