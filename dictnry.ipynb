# Dictionary Methods

| Method         | Syntax                            | Returns       | Purpose                                  | Example                      |
| -------------- | --------------------------------- | ------------- | ---------------------------------------- | ---------------------------- |
| **keys()**     | `dict.keys()`                     | View object   | Returns all keys                         | `d.keys()`                   |
| **values()**   | `dict.values()`                   | View object   | Returns all values                       | `d.values()`                 |
| **items()**    | `dict.items()`                    | View object   | Returns `(key, value)` pairs             | `d.items()`                  |
| **get()**      | `dict.get(key, default=None)`     | Value         | Gets value without raising `KeyError`    | `d.get("name")`              |
| **update()**   | `dict.update(other_dict)`         | `None`        | Adds or updates key-value pairs          | `d.update({"age":25})`       |
| **pop()**      | `dict.pop(key, default)`          | Removed value | Removes a specified key                  | `d.pop("age")`               |
| **popitem()**  | `dict.popitem()`                  | Tuple         | Removes the last inserted item           | `d.popitem()`                |
| **clear()**    | `dict.clear()`                    | `None`        | Removes all items                        | `d.clear()`                  |
| **copy()**     | `dict.copy()`                     | Dictionary    | Creates a shallow copy                   | `new_d = d.copy()`           |
| **fromkeys()** | `dict.fromkeys(keys, value=None)` | Dictionary    | Creates a new dictionary with given keys | `dict.fromkeys(["a","b"],0)` |


Here's a 5 usefull methods

| Method         | Remember As                        | Syntax                              | Example               |
| -------------- | ---------------------------------- | ----------------------------------- | --------------------- |
| **`keys()`**   | Returns all **keys**               | `dictionary.keys()`                 | `student.keys()`      |
| **`values()`** | Returns all **values**             | `dictionary.values()`               | `student.values()`    |
| **`items()`**  | Returns **key-value pairs**        | `dictionary.items()`                | `student.items()`     |
| **`get()`**    | **Safely gets** the value of a key | `dictionary.get(key, default=None)` | `student.get("name")` |

# Add , Modify items in a dictionary

```python
# add item using assign '=' operator
student = {
    1 : 'Class-X',
    'name' : 'Ajay',
    'grade' : 'A',
    'city' : 'Hyderabad'
}
student['email_id'] = 'a1@gmail.com'
print(student)
```

```python
# Modify or Replace the values using assign '=' operator
student = {
    1 : 'Class-X',
    'name' : 'Dharneesh',
    'grade' : 'A',
    'city' : 'Chennai'
}
student['grade'] = 'S+'
print(student)
```

```python
# Remove items Method 1
# del to remove items
student = {
    1 : 'Class-X',
    'name' : 'Dharneesh',
    'grade' : 'A',
    'city' : 'Chennai'
}
del student['grade']
print(student)
```

```python
# Remove items Method 2
# pop to remove items 
student = {
    1 : 'Class-X',
    'name' : 'Dharneesh',
    'grade' : 'A',
    'city' : 'Chennai'
}
var1 = student.pop('city')
print(var1)
print(student)
```

# Dictionary Itterationsing for Loop

A dictionary can be itterated using for loop. we can loop through dictionaries by keys, values or both

# Loop through keys

```python
student = {
    1 : 'Class-X',
    'name' : 'Ajay',
    'grade' : 'A',
    'city' : 'Hyderabad'
}
for keys in student:
    print(keys)
```

# loop through the values

```python
# method 1 using .value()
student = {
    1 : 'Class-X',
    'name' : 'Ajay',
    'grade' : 'A',
    'city' : 'Hyderabad'
}
for value in student.values():
    print(value)
```

```python
# method 2
student = {
    1 : 'Class-X',
    'name' : 'Ajay',
    'grade' : 'A',
    'city' : 'Hyderabad'
}
for value in student:
    print(student[value])
```

# Loop through key value pair using item()

```python
student = {
    1: 'Class-X', #use numberic datatype
    'name' : 'Virat',
    'grade' : 'A',
    'city' : 'Bangalore'
}
for keys,value in student.items():
    print(keys,value)
```

# Dictionary Comprehension

## Syntax :

#### new_dic = {key_exp : value:exp for item in itterable if condition(optional)}

```python
my_dic = {x : x*x for x in range(1,6)}
print(my_dic)
```

```python
my_dic = {x : x+x for x in range(1,6)}
print(my_dic)
```
